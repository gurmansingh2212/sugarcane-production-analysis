```python
import pandas as pd
import seaborn as sns
from matplotlib import pyplot as plt
```

```python
df = pd.read_csv("List of Countries by Sugarcane Production.csv")
```

```python
df.head()
```

```python
df.shape
```

**Data** **Cleaning**

```python
df.rename(columns = {"Production (Tons)": "Production(Tons)"}, inplace= True)
df.rename(columns = {"Production per Person (Kg)": "Production_per_Person(Kg)"}, inplace= True)
df.rename(columns = {"Acreage (Hectare)": "Acreage(Hectare)"}, inplace= True)
df.rename(columns = {"Yield (Kg / Hectare)": "Yield(Kg/Hectare)"}, inplace= True)
```

```python
df.head()
```

```python
df["Production(Tons)"] = df["Production(Tons)"].str.replace(".","")
df["Production_per_Person(Kg)"] = df["Production_per_Person(Kg)"].str.replace(".","").str.replace(",",".")
df["Acreage(Hectare)"] = df["Acreage(Hectare)"].str.replace(".","")
df["Yield(Kg/Hectare)"] = df["Yield(Kg/Hectare)"].str.replace(".","").str.replace(",",".")
```

```python
df.head()
```

```python
df.isna().sum()
```

```python
df[df["Acreage(Hectare)"].isnull()]
```

```python
df.dtypes
```

```python
df["Production(Tons)"] = df["Production(Tons)"].astype(float)
df["Production_per_Person(Kg)"] = df["Production_per_Person(Kg)"].astype(float)
df["Acreage(Hectare)"] = df["Acreage(Hectare)"].astype(float)
df["Yield(Kg/Hectare)"] = df["Yield(Kg/Hectare)"].astype(float)
```

```python
df.head(10)
```

**Which country has highest land**

```python
df_acr = df.sort_values("Acreage(Hectare)", ascending = False)
ax = sns.barplot(data = df_acr.head(10), x = "Country", y = "Acreage(Hectare)")
ax.set_xticklabels(ax.get_xticklabels(), rotation = 90)
plt.show()
```

**Which country has highest yield per hectare**

```python
df_yield = df.sort_values("Yield(Kg/Hectare)", ascending = False)
ax = sns.barplot(data = df_yield.head(10), x = "Country", y = "Yield(Kg/Hectare)")
ax.set_xticklabels(ax.get_xticklabels(), rotation = 90)
plt.show()
```

**Which country has the highest production**

```python
df_production = df.sort_values("Production_per_Person(Kg)", ascending = False)
ax = sns.barplot(data = df_production.head(10), x = "Country", y = "Production_per_Person(Kg)")
ax.set_xticklabels(ax.get_xticklabels(), rotation = 90)
plt.show()
```

**Analysis for continent**

```python
df_continent = df.groupby("Continent").sum()
```

```python
df_continent
```

**Continent produces maximum sugarcane**

```python
df_continent["Production(Tons)"].sort_values(ascending = False).plot(kind = "bar")
```

```python
df_continent["Acreage(Hectare)"].sort_values(ascending = False).plot(kind = "bar")
```

No. of countries in a continent effects production of sugarcane

```python
df_continent["number_of_countries"] = df.groupby("Continent").count()["Country"]
```

```python
df_continent
```

```python
continent_names = df_continent.index.to_list()
sns.lineplot(data = df_continent, x = "number_of_countries", y = "Production(Tons)")
plt.xticks(df_continent["number_of_countries"], continent_names , rotation = 90)
plt.show()
```
